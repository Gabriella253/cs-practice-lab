````python
class Node:
  def __init__(self, data):
    self.data = data
    self.next = None

n1 = Node("A")
n2 = Node("B")
n3 = Node("C")
````

````python
# EXERCÍCIO 2 — Construindo uma cadeia

# Conectando os nós criados acima: A -> B -> C -> None
n1.next = n2
n2.next = n3

# Percorrendo e imprimindo a estrutura
atual = n1
while atual is not None:
  print(atual.data)
  atual = atual.next
````

### Exercício 3 — Depuração

1. **Qual é o problema?**
   O programa entra em um loop infinito.

2. **Por que o programa não termina?**
   Porque a variável `atual` nunca avança para o próximo nó dentro do bloco `while`. Ela fica presa referenciando o nó `n1` para sempre, mantendo a condição do loop verdadeira por tempo indeterminado.

3. **Qual linha deve ser acrescentada?**
   Deve ser adicionado o avanço de ponteiro `atual = atual.proximo` ao final do loop.

4. **Qual será a saída depois da correção?**
   ```text
   10
   20
   30
   ```

````python
# Código Corrigido do Exercício 3

# Nota: Adaptado para a classe Node padrão proposta no enunciado (valor/proximo)
class NodeDepuracao:
  def __init__(self, valor):
    self.valor = valor
    self.proximo = None

n1_dep = NodeDepuracao(10)
n2_dep = NodeDepuracao(20)
n3_dep = NodeDepuracao(30)

n1_dep.proximo = n2_dep
n2_dep.proximo = n3_dep

atual = n1_dep
while atual is not None:
  print(atual.valor)
  atual = atual.proximo  # Linha acrescentada para resolver o problema
````
