# Atividade 1 Listas

Crie uma lista com 10 números inteiros, solicite os valores ao usuário, exiba os números, calcule a soma, a média, o maior valor, o menor valor e a quantidade de números pares.

```python
opcao = input("Digite sua opção\n Opcao 1 - Programa\n Opcao 2 - Sair\n")
print(opcao)

if opcao == "1":
  entrada = input("Digite 10 números separados por espaço: ")
  numeros = entrada.split()
  numeros_inteiros_map = list(map(int, numeros))

  if len(numeros_inteiros_map) != 10:
    print("Você deve digitar exatamente 10 números.")

  else:
    print("Números digitados pelo usuário:", numeros_inteiros_map)

    numeros_somados = sum(numeros_inteiros_map)
    print("A soma dos números é:", numeros_somados)

    media = numeros_somados / len(numeros_inteiros_map)
    print("A média dos números é:", media)

    maior = max(numeros_inteiros_map)
    print("O maior valor é:", maior)

    menor = min(numeros_inteiros_map)
    print("O menor valor é:", menor)

    q_num_pares = 0

    for num in numeros_inteiros_map:
      if num % 2 == 0:
        q_num_pares += 1

    print("A quantidade de números pares é:", q_num_pares)

else:
  print("Opção inválida\nEncerrando programa.")
```

# Atividade 2 Matrizes

Crie uma matriz 3 × 3, solicite os valores ao usuário e calcule a soma de todos os elementos, a soma da diagonal principal e o maior elemento.

```python
matriz = []

for i in range(3):
  linha = []

  for j in range(3):
    numero = int(input(f"Digite o valor da posição [{i}][{j}]: "))
    linha.append(numero)

  matriz.append(linha)

print("\nMatriz:")

for linha in matriz:
  print(*linha)

soma = 0
soma_diagonal = 0
maior = matriz[0][0]

for i in range(3):
  for j in range(3):
    soma += matriz[i][j]

    if i == j:
      soma_diagonal += matriz[i][j]

    if matriz[i][j] > maior:
      maior = matriz[i][j]

print("\nA soma de todos os elementos é:", soma)
print("A soma da diagonal principal é:", soma_diagonal)
print("O maior elemento é:", maior)
```

# Atividade 4 Dataclass

Crie uma dataclass chamada Aluno com nome, matrícula, nota1 e nota2. Crie um método media() para calcular a média das duas notas.

```python
from dataclasses import dataclass

@dataclass
class Aluno:
  nome: str
  matricula: int
  nota1: float
  nota2: float

  def media(self):
    return (self.nota1 + self.nota2) / 2


aluno = Aluno(
  "Maria",
  12345,
  8.0,
  9.0
)

print("Nome:", aluno.nome)
print("Matrícula:", aluno.matricula)
print("Nota 1:", aluno.nota1)
print("Nota 2:", aluno.nota2)
print("Média:", aluno.media())
```

# Atividade 5 Lista de objetos

Utilizando a dataclass Aluno, cadastre cinco alunos. Calcule a média de cada aluno, exiba os aprovados e identifique o aluno com maior média.

```python
from dataclasses import dataclass

@dataclass
class Aluno:
  nome: str
  matricula: int
  nota1: float
  nota2: float

  def media(self):
    return (self.nota1 + self.nota2) / 2


alunos = []

for i in range(5):
  print(f"\nCadastro do {i + 1}º aluno")

  nome = input("Nome: ")
  matricula = int(input("Matrícula: "))
  nota1 = float(input("Nota 1: "))
  nota2 = float(input("Nota 2: "))

  aluno = Aluno(nome, matricula, nota1, nota2)
  alunos.append(aluno)


print("\n--- Médias dos alunos ---")

for aluno in alunos:
  print(f"{aluno.nome}: {aluno.media():.2f}")


print("\n--- Alunos aprovados ---")

for aluno in alunos:
  if aluno.media() >= 7.0:
    print(f"{aluno.nome} - Média: {aluno.media():.2f}")


aluno_maior_media = alunos[0]

for aluno in alunos:
  if aluno.media() > aluno_maior_media.media():
    aluno_maior_media = aluno


print("\n--- Aluno com maior média ---")
print("Nome:", aluno_maior_media.nome)
print("Média:", f"{aluno_maior_media.media():.2f}")
```

# Desafio integrador — Sistema de estoque

Crie um sistema de estoque utilizando uma dataclass Produto. O programa permite cadastrar produtos, listar produtos, calcular o valor total do estoque, localizar um produto pelo código e identificar o produto com maior preço.

```python
from dataclasses import dataclass

@dataclass
class Produto:
  codigo: int
  nome: str
  preco: float
  quantidade: int


produtos = []


def cadastrar_produto():
  print("\n--- Cadastro de produto ---")

  codigo = int(input("Código: "))
  nome = input("Nome: ")
  preco = float(input("Preço: "))
  quantidade = int(input("Quantidade: "))

  produto = Produto(codigo, nome, preco, quantidade)
  produtos.append(produto)

  print("Produto cadastrado com sucesso!")


def listar_produtos():
  print("\n--- Produtos cadastrados ---")

  if len(produtos) == 0:
    print("Nenhum produto cadastrado.")
    return

  for produto in produtos:
    print(f"Código: {produto.codigo}")
    print(f"Nome: {produto.nome}")
    print(f"Preço: R$ {produto.preco:.2f}")
    print(f"Quantidade: {produto.quantidade}")
    print()


def calcular_valor_estoque():
  total = 0

  for produto in produtos:
    total += produto.preco * produto.quantidade

  print(f"\nValor total do estoque: R$ {total:.2f}")


def localizar_produto():
  codigo = int(input("\nDigite o código do produto: "))

  for produto in produtos:
    if produto.codigo == codigo:
      print("\nProduto encontrado:")
      print(f"Código: {produto.codigo}")
      print(f"Nome: {produto.nome}")
      print(f"Preço: R$ {produto.preco:.2f}")
      print(f"Quantidade: {produto.quantidade}")
      return

  print("Produto não encontrado.")


def produto_maior_preco():
  if len(produtos) == 0:
    print("\nNenhum produto cadastrado.")
    return

  maior = produtos[0]

  for produto in produtos:
    if produto.preco > maior.preco:
      maior = produto

  print("\n--- Produto com maior preço ---")
  print(f"Código: {maior.codigo}")
  print(f"Nome: {maior.nome}")
  print(f"Preço: R$ {maior.preco:.2f}")
  print(f"Quantidade: {maior.quantidade}")


while True:
  print("\n========== SISTEMA DE ESTOQUE ==========")
  print("1 - Cadastrar produto")
  print("2 - Listar produtos")
  print("3 - Calcular valor total do estoque")
  print("4 - Localizar produto pelo código")
  print("5 - Produto com maior preço")
  print("6 - Sair")

  opcao = input("Escolha uma opção: ")

  if opcao == "1":
    cadastrar_produto()

  elif opcao == "2":
    listar_produtos()

  elif opcao == "3":
    calcular_valor_estoque()

  elif opcao == "4":
    localizar_produto()

  elif opcao == "5":
    produto_maior_preco()

  elif opcao == "6":
    print("Programa encerrado.")
    break

  else:
    print("Opção inválida.")
```
