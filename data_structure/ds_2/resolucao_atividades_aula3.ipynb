```python
#SISTEMA DE FILA DE CLÍNICA / RESOLUÇÃO DA SITUAÇÃO PROBLEMA

filapacientes = []
filaprioridade = []
#alterar o código depois pra adicionar uma segunda fila de prioridade 60+ de acordo com a idade
opcao = True

while opcao == True:
  nome = input("Insira o nome do paciente: ")
  idade = input("Insira a idade do paciente: ")
  situacao = input("Insira a situação do paciente:\n1- Pode esperar (20 a 30m)\n2- Precisa de atendimento (5 a 10m)\n3- Emergência (atendimento imediato)\n")

  if situacao == "1":
    print("Situação do paciente: Pode esperar")
    filapacientes.append(nome)
    print("Situação atual da fila: ", filapacientes)
    opcao2 = input("Deseja inserir outro paciente? 1- Sim 2- Não\n")
    if opcao2 == "1":
        opcao = True
    elif opcao2 == "2":
        opcao = False
        print("Encerrando programa.")
    #acho que daria pra alterar esses ifs por uma função mas prefiro não usar agora

  elif situacao == "2":
    print("Situação do paciente: Precisa de atendimento")
    filapacientes.insert(1, nome)
    print("Situação atual da fila: ", filapacientes)
    opcao2 = input("Deseja inserir outro paciente? 1- Sim 2- Não\n")
    if opcao2 == "1":
        opcao = True
    elif opcao2 == "2":
        opcao = False
        print("Encerrando programa.")

  elif situacao == "3":
    print("Situação do paciente: Emergência")
    filapacientes.insert(0, nome)
    print("Situação atual da fila: ", filapacientes)
    opcao2 = input("Deseja inserir outro paciente? 1- Sim 2- Não\n")
    if opcao2 == "1":
        opcao = True
    elif opcao2 == "2":
        opcao = False
        print("Fila: ", filapacientes)
        print("Encerrando programa.")
```
