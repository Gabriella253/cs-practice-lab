```python
class EditorTexto:
    # o construtor aqui ele prepara tudo pro editor
    def __init__(self):
        # essa eh a lista que guarda o historico pra poder desfazer
        self.pilha_desfazer = []
        # aqui a gente guarda o que ta escrito agora
        self.texto_atual = ""

    # essa funcao faz as acoes tipo digitar apagar ou trocar o texto
    def ejecutar_acao(self, acao, conteudo):
        # antes de fazer qualquer coisa a gente salva o texto atual pra poder voltar depois
        self.pilha_desfazer.append(self.texto_atual)

        # se for pra digitar ele adiciona o conteudo
        if acao == "digitar":
            self.texto_atual += conteudo
        # se for pra apagar ele tira uns caracteres do final
        elif acao == "apagar":
            # transforma o conteudo em numero porque e a quantidade de caracteres
            num_caracteres = int(conteudo)
            self.texto_atual = self.texto_atual[:-num_caracteres]
        # se for pra substituir ele troca tudo pelo conteudo novo
        elif acao == "substituir":
            self.texto_atual = conteudo

        # imprime o que foi feito e como ficou o texto
        print(f"Ação: {acao} '{conteudo}' -> Texto atual: \"{self.texto_atual}\"")

    # essa funcao desfaz a ultima acao
    def desfazer(self):
        # se nao tiver nada na pilha pra desfazer ele avisa
        if not self.pilha_desfazer:
            print("Nada para desfazer.")
            return

        # ele pega o ultimo texto salvo na pilha e coloca como atual
        self.texto_atual = self.pilha_desfazer.pop()
        # avisa que desfez e qual texto voltou
        print(f"Desfazer acionado! -> Texto restaurado: \"{self.texto_atual}\"")


# cria uma instancia do nosso editor de texto
editor = EditorTexto()

# aqui a gente faz umas acoes no editor pra testar
editor.ejecutar_acao("digitar", "Olá ")
editor.ejecutar_acao("digitar", "Mundo")
editor.ejecutar_acao("apagar", "5")
editor.ejecutar_acao("substituir", "Olá Universo")

# avisa que vai comecar a desfazer
print("\n--- Iniciando os comandos de Desfazer ---")
# desfaz tres vezes pra ver o historico voltando
editor.desfazer()
editor.desfazer()
editor.desfazer()
```

```python
class SpoolerImpressao:
    def __init__(self):
        # cria a lista vazia pra guardar os docs
        self.fila = []

    def adicionar_documento(self, nome_documento):
        # coloca o documento novo bem no fim da fila
        self.fila.append(nome_documento)
        print(f"Documento '{nome_documento}' adicionado à fila.")

    def imprimir_documento(self):
        # confere se tem algo pra imprimir senao nao faz nada
        if self.esta_vazia():
            print("A fila de impressão está vazia.")
            return None

        # o pop zero tira o primeirao que entrou da lista (regrinha do fifo)
        documento = self.fila.pop(0)
        print(f"A imprimir: {documento}")
        return documento

    def esta_vazia(self):
        # se o tamanho for zero significa q ta vazia
        return len(self.fila) == 0

    def exibir_fila(self):
        if self.esta_vazia():
            print("Nenhum documento aguardando.")
        else:
            # mostra a fila bonitinha com as setas juntando tudo
            print("Fila atual de impressão:", " -> ".join(self.fila))


if __name__ == "__main__":
    spooler = SpoolerImpressao()

    # adicionando os testes pra rodar
    spooler.adicionar_documento("Relatorio_Financeiro.pdf")
    spooler.adicionar_documento("Contrato_Final.docx")
    spooler.adicionar_documento("Foto_Grupo.png")

    spooler.exibir_fila()

    print("\n--- Iniciando Impressão ---")
    # vai tirando um por um na ordem certa
    spooler.imprimir_documento()
    spooler.imprimir_documento()

    print("\n--- Estado Atual ---")
    spooler.exibir_fila()
```

```python
from collections import deque

class TriagemHospitalar:
    # o construtor pra comecar a fila do hospital
    def __init__(self):
        # Inicializa a fila dupla (deque) que eh boa pra adicionar dos dois lados
        self.fila = deque()

    # pra adicionar paciente que nao tem pressa
    def adicionar_paciente_normal(self, nome):
        # ele vai pro final da fila normal
        self.fila.append(nome)
        print(f"Paciente normal '{nome}' adicionado ao final da fila.")

    # pra adicionar paciente que tem pressa
    def adicionar_paciente_prioritario(self, nome):
        # esse fura a fila e vai pro comeco
        self.fila.appendleft(nome)
        print(f"⚠️ Paciente prioritário '{nome}' inserido no início da fila!")

    # pra chamar o proximo pra ser atendido
    def proximo_atendimento(self):
        # se tiver gente na fila ele chama
        if self.fila:
            # Remove e retorna o proximo paciente que ta no comeco da fila
            paciente = self.fila.popleft()
            print(f"A chamar para atendimento: {paciente}")
            return paciente
        # se nao tiver ninguem ele avisa
        print("A fila está vazia.")
        return None

    # pra mostrar quem ta na fila agora
    def mostrar_fila(self):
        print("Estado atual da fila:", list(self.fila))

# Exemplo de utilização:
hospital = TriagemHospitalar()

# adicionando pacientes normais
hospital.adicionar_paciente_normal("João")
hospital.adicionar_paciente_normal("Maria")
hospital.mostrar_fila()

# Chega um paciente prioritário (ex: emergência ou idoso)
hospital.adicionar_paciente_prioritario("Carlos (Emergência)")
hospital.mostrar_fila()

# Chamando para atendimento
hospital.proximo_atendimento()
hospital.mostrar_fila()
```
